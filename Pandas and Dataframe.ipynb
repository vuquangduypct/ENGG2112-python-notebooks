### Import data from an excel file

'pd.read_excel()' function is used to import data in an excel file into a special data type called a dataframe. The default setting is to read the data in the first sheet.The details of the function can be found in: https://pandas.pydata.org/docs/reference/api/pandas.read_excel.html

'pd.read_csv()' function is used when the file is 'csv' type instead of the 'excel' type.The details of the function can be found in: https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html

By default, the first row of the CSV or XLS file is taken to be a header row. If there is no header, the keyword argument header=None should be provided to read_excel and read_csv.

```python
excel_file = 'movies.xls'
movies = pd.read_excel(excel_file,index_col=0)
# index_col=0 indicates that there should not be a column of running values inserted. By default, there would be.
```

Show the created dataframe.

```python
movies
```

df.head() and df.tail() functions return the first 5 rows and the last 5 rows of a dataframe, respectively. These two functions are normally used to show 5 lines of the dataframe as an example, instead of showing the whole dataframe, expecially when the dataframe is large.

```python
movies.head()
```

```python
movies.tail()
```

Read data in a specific sheet by using 'sheet_name' feature

```python
movies_sheet1 = pd.read_excel(excel_file, sheet_name=0, index_col=0)
movies_sheet2 = pd.read_excel(excel_file, sheet_name=1, index_col=0)
movies_sheet3 = pd.read_excel(excel_file, sheet_name=2, index_col=0)
```

### Merge two dataframes into one

pd.concat() is a powerful function to merge multiple dataframes together. A list of dataframes is required in this function.

For example, importing data in all sheets in a loop, and append them in a list of dataframes. Using pd.concat(list()) to merge them together.

```python
movie_sheets = []
movie_sheets.append(movies_sheet1)
movie_sheets.append(movies_sheet2)
movie_sheets.append(movies_sheet3)
movies_db = pd.concat(movie_sheets)
movies_db.tail() # This would be the same as the last few entries of the 3rd sheet in the original Excel file
```

'shape' is the method to extract the size (no. of rows and columns) of a dataframe.

```python
movies_db.shape
```

### Remove samples with missing values

Missing values are presented as 'NaN' in dataframes. Sometimes, missing values need to be removed in a dataframe for further operations. 

Here are two ways to remove missing values.

1. 'df.dropna(how='')' 

function can remove samples with missing values. when how='all', the function remove the rows with 'NaN' in all columns. When how='any', the function remove the rows with 'NaN' in any column. While removing all rows with 'NaN' in a specific column, 'subset' is involved. 
For example, remove all samples with missing values of 'Gross Earnings' in the dataframe movies.

```python
movies_removeNaN1 = movies_db.dropna(subset=["Gross Earnings"])
```

```python
movies_removeNaN1.shape
```

The second function

2. df[' '].notna()

can also remove rows with missing values in a specific column.

```python
movies_removeNaN2 = movies_db[movies_db['Gross Earnings'].notna()]
```

```python
movies_removeNaN2.shape
```

'movies_removeNaN1' and 'movies_removeNaN2' dataframes have the same shape as (4159, 24), which indicates both functions work the same in this case.

### Search in a dataframe

'df.loc\[]' is a powerful function to fetch an expected data record in a dataframe. This function can implement a searching operation. For example, searching all data records with 'IMDB score' over 8.0. The result will return a dataframe with all records satisfying the search criteria. 

```python
movies.loc[movies['IMDB Score'] > 8.0]
```

Similarly, we seach all data records categorised as 'Drama' genre.

```python
movies.loc[movies['Genres'].str.contains("Drama", case=False)]
```

### Sort a dataframe

'df.sort_values()' function is used to sort a dataframe by ascending or discending order based on a specific feature (column). The default setting is 'ascending=True' with an ascending order while sorting in discending order when 'ascending=False'.

For example, sorting 'Gross Earnings' in the dataframe imported above with a discending order.

```python
sorted_by_gross = movies.sort_values(['Gross Earnings'], ascending=False)
```

Show the first 10 records of sorted dataframe.

```python
sorted_by_gross['Gross Earnings'].head(10)
```

## Data Visualization

Import package 'matplotlib' to plot data in required type of charts (histogram, line graph, bar chart and pie chart).
Dataframe supports a direct plot by using df.plot() function.

```python
import matplotlib.pyplot as plt
```

### bar chart

For example, plotting the first 10 records of above sorted results in a vertical bar chart. The default setting of the function df.plot() simply shows the index of shown data and their values. 

```python
sorted_by_gross['Gross Earnings'].head(10).plot(kind="bar")
plt.show()
```

### Horrizon bar chart

Plot the first 10 records of above sorted results in a horrizon bar chart. x-axis is , and y-axis is set as movie names. A chart title and legend are also included in the plot to make chart clearer.

```python
sorted_by_gross.head(10).plot(x='Title', y='Gross Earnings', kind="barh")
plt.title('Top ten movies in Gross Earnings') 
plt.legend(loc='upper right')
plt.show()
```

### Histograms

Plot the histogram chart for the frequency of 'IMDB Score' in the dataframe. Also include axis labels this time ('Frequency' for the y-axis, and 'Score' for the x-axis.). 

```python
movies['IMDB Score'].plot(kind="hist")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title('Frequency of IMDB Score')
plt.show()
```

### Line graph

Line chart is with the kind as 'line'. While the title of movies may be too long to demonstrate as a label of x-axis, 'plt.xticks(rotation)' function can be used to adjust the angle of label, for example 'rotation = -30' means the angle is adjusted as -30 degree.

```python
sorted_by_gross.head().plot(x='Title', y='Gross Earnings', kind="line")
plt.xticks(rotation=-30)
plt.show()
```
