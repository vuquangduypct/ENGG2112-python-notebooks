### Array

NumPy contains N-dimensional array objects called "ndarray", which is a collection of data of the same type. It functions as a highly efficient list for multiple pieces of data.

Change list to array in numpy



```python
data1 = [1,2,3,4,5]
array1 = np.array(data1)
print(array1)
print(type(array1))
print("Each element is of type",array1.dtype) # The dtype method reveals the data type of each element.
```

A nested list can be converted into a multi-dimensional array, called a matrix.



```python
data2 = [[1,2,3],[3,4,5]]
array2 = np.array(data2)
print(array2)
print("Shape:", array2.shape) # The shape method reveals the dimensions of the ndarray.
```

Array calculation in NumPy is very convenient. You can apply operations to the whole array at the same time.



```python
array1
```

```python
[array1 + 1, array1*2] # Element-wise addition and multiplication by a constant.
```

```python
array2 = np.array([2,3,4,5,6])
array1 * array2 # Element-wise multiplication of two arrays with the same dimensions
```

```python
print(array1 ** 2) # Element-wise squaring
```

### Matrix Multiplications - Matmul
Quite often we need to multiply matrices and NumPy provides the function matmul to accomplish this. Note that the * operation performs element-wise multiplication only.

```python
A = np.array([[1, 0],
              [0, 2]])
B = np.array([[4, 1],
              [2, 2]])
np.matmul(A, B) # Computes the matrix AB. Ordering matters in matrix multiplication, so np.matmul(B,A) != np.matmul(A,B)
                # in general.
```

## Generate Random Numbers
In simulations, it is useful to generate random numbers in a controlled manner so that we can have a large amount of data to work with. The random values you would most likely encounter are drawn from Gaussian (or normal) and uniform distributions. Python has several ways to generate these.

### Uniform

```python
import numpy as np
# This generates a random value uniformly distributed between 0 (low) and 1 (high). The size input argument can be used
# to specify the dimensions of an ndarray of uniform random numbers.
X1 = np.random.uniform(low=0.0, high=1.0, size=None)
X2 = np.random.uniform(low=0, high = 10, size = (2,3))
print("Here's one random value: ", X1)
print("Here's a matrix of random values between 0 and 10:\n", X2)
```

### Gaussian

```python
# Here we generate a normally distributed random number with mean given by loc and variance by scale. As above, we can
# also generate a matrix of random numbers.
X3 = np.random.normal(loc=0.0, scale=1.0, size=None)
X4 = np.random.normal(loc=1.0, scale = 1.0, size=(2,3))
print("Here's one N(0,1) random value: ", X3)
print("Here's a matrix of N(1,1) random values:\n", X4)
```

### Uniform integers

```python
# The randint function generates a uniform integer-valued random number between 0 and k-1.
X5 = np.random.randint(100)
X6 = np.random.randint(100,size=(2,3))
print("Here's one integer random value: ", X5)
print("Here's a matrix of integer random values:\n", X6)
```

### BitGenerator Generator
To create a repeatable sequence of "random" numbers, we use the default_rng method with a generating seed as input. This yields the same sequence of numbers every time.

```python
rng = np.random.default_rng(1) # The input argument is the seed, which can be any non-negative integer value.
rng.random(size=(1,100)) # This will produce exactly the same 100 "random" numbers if the preceding line is executed.
```

## Saving Into/Loading From Files
We would often need to save the outputs of our simulations so that we don't have to keep re-running them, as that may take hours or in more complex cases, days. Numpy arrays can be stored into a special .npy or .npz format using the save or savez functions, and read from these types of files using the load function.

We can also save/load arrays in text format using the savetxt and loadtxt functions. Doing this allows us to read the files outside of numpy, e.g. using pandas, if the file is saved in a CSV format.

```python
# Let's try to save X1, X2 and X3 into a file using the savez function (which saves multiple variables at the same time)
outfile = "Example1.npz" # We can also provide the full or relative path name if the file should not be in the current
                         # directory.
np.savez(outfile,X1=X1,X2=X2,X3=X3) # Need to provide savez with keyword arguments, otherwise the variables are named
                                    # 'arr_0', 'arr_1', etc.

# Now load them back to verify that we can recover our data...
npzfile = np.load(outfile) # This is an object that stores all the variables in the file.
print(npzfile.files) # This reveals the names of the variables within the file.
Y1 = npzfile['X1'] # The npzfile object contains the data as a dictionary, the elements of which can be referenced by
                   # their dictionary labels.
print(X1,Y1) # To show that Y1 = X1.
Y2 = npzfile['X2']
print(X2 == Y2)
```

```python
# Here we store the array X4 into a comma-delimited file and read it using pandas (instead of numpy)
import pandas as pd

np.savetxt("Example2.csv",X4,delimiter=',')
Y4 = pd.read_csv("Example2.csv",header=None) # There's no header row in this CSV file, so header=None is needed.
Y4.to_numpy() # As shown in the pandas tutorial, this step converts the pandas data frame into a numpy array
```

# Line Plot
You would have encountered x-y plots many times in the past, e.g. in plotting y = sin(x) and other functions. In engineering, we are often interested in signals that are functions of time, and these can be visualised using an x-y plot with x as the time variable (usually t).

To create a line graph, we use the <strong>plt.plot ()</strong> function.

Add a title for the line plot and a label the x and y axes. Then run the code to see your graph.

<strong>Note:</strong> In the code below, one dataset has been assigned the label 'x', and another to the label 'y'. These labels do not correspond to an axis, so we can actually name these data sets anything we like. The first label you name in the plt.plot() function will form your x (horizontal) axis while the second will be your y (vertical) axis. Try switching the x and y labels in the plt.plot() function and run the code to see what happens.

```python
x = np.arange(1,6)
y = np.random.random(5)
plt.plot(x,y) # Plot y against x. x and y must have the same length (no. of elements)
plt.title('My first plot') # This puts a title at the top
plt.xlabel('x') # Label for the horizontal axis
plt.ylabel('y') # Label for the vertical axis
plt.grid() # Add grid lines

plt.show() # Sometimes this is needed for the plot to show up nicely.
```

```python
# You can also change the size of your figure. figsize=(xlen,ylen) will indicate an x-width of xlen and a y-height of ylen.
fig1 = plt.figure(figsize=(10,10))
plt.plot(x,y)
plt.show(fig1) # Show fig1 first...

# Multiple figures can be created and the plt.show() function will show the figure referenced in its input argument.
fig2 = plt.figure(figsize=(10,5))
plt.plot(2*x,y/2)
plt.grid(fig2)
plt.show(fig2)
```

```python
# We would often want to plot multiple lines in one plot. This is done using multiple calls of plt.plot while working
# on the same figure, or by giving multiple (x,y) input arguments to plt.plot.
x1 = range(10)
y1 = np.random.normal(size=(10,))
x2 = range(10)
y2 = np.random.normal(size=(10,))
# plt.plot(x1,y1)  # Uncomment these two lines to see that they produce the same result as the line below.
# plt.plot(x2,y2)

plt.plot(x1,y1,x2,y2)
plt.grid()
```

```python
# Multiple lines on one plot can (and should!) be named in a legend.
plt.plot(x1,y1,label='curve 1')
plt.plot(x2,y2,label='curve 2')
plt.grid()
plt.legend()
plt.show()
```

```python
# We can save our figure into any file format
plt.plot(x1,y1,label='curve 1')
plt.plot(x2,y2,label='curve 2')
plt.grid()
plt.legend()
plt.savefig('Fig1.jpg') # The file extension can be PNG, EPS, JPG, etc. A full or relative path can be provided if
                        # the file is not to be saved in the current directory.
```
