## Background

When running ML experiments, you need to track what you tried and how well it worked. In this task you will build a simple `ExperimentTracker` class and use it to log results, then visualise them.

Each experiment record stores:

| Field | Type | Description |
|---|---|---|
| `name` | `str` | A short label for the run |
| `model` | `str` | Model architecture or algorithm |
| `accuracy` | `float` | Validation accuracy (0–1) |
| `f1` | `float` | Macro F1 score (0–1) |
| `epochs` | `int` | Number of training epochs |

### Evaluation metric reminder

The **F1 score** is the harmonic mean of precision $P$ and recall $R$:

$$F_1 = 2 \cdot \frac{P \cdot R}{P + R}$$

We use it alongside accuracy because accuracy alone can be misleading on imbalanced datasets.

## Step 4 — Visualise results with `matplotlib` 📊

A key notebook feature is **inline plotting**. The cell below creates a bar chart comparing accuracy and F1 across all runs.

Read through the code, then run it. After it renders, answer the question in the markdown cell that follows.

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.dpi'] = 110

names     = [exp.name     for exp in tracker.runs]
models    = [exp.model    for exp in tracker.runs]
accuracy  = [exp.accuracy for exp in tracker.runs]
f1_scores = [exp.f1       for exp in tracker.runs]

x = range(len(names))
width = 0.38

fig, ax = plt.subplots(figsize=(11, 4.5))

bars_acc = ax.bar([i - width/2 for i in x], accuracy,  width, label='Accuracy', color='#0D9488', alpha=0.88)
bars_f1  = ax.bar([i + width/2 for i in x], f1_scores, width, label='F1 Score', color='#F59E0B', alpha=0.88)

# Annotate bars with values
for bar in bars_acc:
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.003,
            f"{bar.get_height():.3f}", ha='center', va='bottom', fontsize=7.5)
for bar in bars_f1:
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.003,
            f"{bar.get_height():.3f}", ha='center', va='bottom', fontsize=7.5)

ax.set_xticks(list(x))
ax.set_xticklabels([f"{n}\n({m[:4]})" for n, m in zip(names, models)], fontsize=8)
ax.set_ylim(0.75, 0.96)
ax.set_ylabel('Score')
ax.set_title('ML Experiment Tracker — Accuracy vs F1 Score by Run', fontsize=12, pad=10)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.4)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
